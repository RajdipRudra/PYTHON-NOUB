<h5 style='text-align:center';><b>Prime Number Checker</div>

```python
def prime_checker(number):
    prime = True
    for i in range(2,number):
        if number%i==0:
            prime = False
    if prime==True:
        print(f"The number {number} is a Prime number")
    else:
         print(f"The number {number} is not a Prime number")
```

```python
n = int(input("Check this number: "))
prime_checker(number=n)
```

```python
def encode(msg,s_num,LL):
    encoded_msg = []
    for i in range(len(msg)):
        letter = msg[i]
        if letter in LL:
            encoded_msg.append(LL[(LL.index(letter))+s_num])
        else:
            encoded_msg.append(letter)
    return "".join(encoded_msg)
```

```python
def decode(msg,s_num,LL):
    decoded_msg = []
    for i in range(len(msg)):
        letter = msg[i]
        if letter in LL:
            decoded_msg.append(LL[(LL.index(letter))-s_num])
        else:
            decoded_msg.append(letter)
    return "".join(decoded_msg)
```
