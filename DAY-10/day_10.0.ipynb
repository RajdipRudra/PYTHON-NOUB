```python

def format_name(f_name,l_name):
    """takes the input string and output a title case format name"""
    return "".join([f_name.title(), l_name.title()])
```

```python
print(format_name("rudra","RAJ"))
```

```python


def leap_year(year):
    leap_year = True
    if year%4==0:
        leap_year = True
        if year%100==0:
            leap_year = False
            if year%400==0:
                leap_year = True
            else:
                leap_year = False
        else:
            leap_year = True
    else:
        leap_year = False

    return leap_year

def days_in_month(year,month):
    month = month-1
    month_days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    leap = leap_year(year=year)
    if leap==True and month==1:
        return 29
    return month_days[month]


user_year = int(input("Which year do you want to check? "))
month = int(input("Enter a month: "))
days = days_in_month(user_year,month)
print(days)
```

```python
def addition(a,b):
    return a+b
def substraction(a,b):
    return a-b
def multi(a,b):
    return a*b
def divition(a,b):
    return a/b



operationss = {
    "+":addition,
    "-":substraction,
    "*":multi,
    "/":divition
}

if "+" in operationss:
    print(True)


```

```python
first_num = int(input("What's the first num?: "))
operation = input("+\n-\n*\n/\npick an operation: ")
sec_num = int(input("what's the next number?: "))

if operation=="+":
    print(f"{first_num}{operation}{sec_num} = {addition(first_num,sec_num)}")
elif operation=="-":
    print(f"{first_num}{operation}{sec_num} = {substraction(first_num,sec_num)}")
elif operation=="*":
    print(f"{first_num}{operation}{sec_num} = {multi(first_num,sec_num)}")
elif operation=="/":
    print(f"{first_num}{operation}{sec_num} = {divition(first_num,sec_num)}")
```
