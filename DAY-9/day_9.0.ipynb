<h5 style='text-align:center';><b>Grading Program</div>

```python
student_scores = {
"Harry": 81,
"Ron": 78,
"Hermione": 99,
"Draco": 74,
"Neville": 62,
}

student_grades = {}
for i in student_scores:
    score = student_scores[i]
    if score>90:
        student_grades[i] = "Outstanding"
    elif score>80:
        student_grades[i] = "EXceeds Expectations"
    elif score>70:
        student_grades[i] = "Acceptable"
    elif score<=70:
        student_grades[i] = "Fail"

 

print(student_grades)
```

<h5 style='text-align:center';><b>Dictionary in List</div>

```python
travel_log = [
    {
        "country": "France",
        "visits": 12,
        "cities": ["Paris", "Lille", "Dijon"]
    },
    {
        "country": "Germany",
        "visits": 5,
        "cities": ["Berlin", "Hamburg", "Stuttgart"]
    }
]

def add_new_country(country,times_V,cities):
    data = {}
    data["country"] = country
    data["visits"] = times_V
    data["cities"] = cities
    travel_log.append(data)

for i in travel_log:
    print(travel_log)
add_new_country("Russia",2,["Moscow","Saint peter'sberg"])
travel_log
```

```python
import os
def winner(dicto):
    win = {}
    price = []
    for i in dicto:
        price.append(dicto[i])
    maximum = max(price)
    print(maximum)
    for j in dicto:
        if dicto[j]==maximum:
            win[j] = maximum
    return win

```

```python
name_price = {}
print("Welcome to the secret auction Program.")

while True:
    name = input("What is your name?-> ")
    bid_price = int(input("What's your bid? -> $"))
    name_price[name]= bid_price
    is_other = input("Are there any other bidders? Type 'yes' or 'no'.\n").lower()
    if is_other=="no" or is_other=="nah" or is_other=="na" or is_other=="n":
        print(winner(name_price))
        break
    os.system("clear")
print(name_price)
```
